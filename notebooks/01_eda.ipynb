Apresenta o catálogo dos 38 datasets de classificação da suíte TabArena-v0.1 (OpenML, suíte 457) e confere a seleção do grupo registrada em `data/selecao_datasets.txt`.

Regra do enunciado: 30 datasets, 10 por regime de tamanho (pequeno: n < 2.500; médio: 2.500 <= n < 10.000; grande: n >= 10.000). Recomendação: pelo menos um problema binário e um multiclasse por regime quando o catálogo oferecer (o validador emite aviso, não erro). O regime médio possui exatamente 10 datasets e, portanto, é comum a todos os grupos.

## Seleção do grupo

```python
selection = read_selection()
report = validate_selection(selection)
for msg in report.warnings:
    print('Aviso:', msg)
for msg in report.errors:
    print('Erro:', msg)
print('Seleção válida.' if report.ok else 'Seleção inválida.')
```

```python
# Tabela para o relatório: nome, task, n, features, classes e regimes.
summarize(selection)
```
